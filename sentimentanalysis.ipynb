```python
import warnings
from langchain._api import LangChainDeprecationWarning

warnings.simplefilter("ignore", category=LangChainDeprecationWarning)
```

```python
import os
from dotenv import load_dotenv,find_dotenv
_ = load_dotenv(find_dotenv())

groq_api_key =  os.environ["GROQ_API_KEY"]
```

```python
from langchain_groq import ChatGroq
llmModel = ChatGroq(temperature=0,model = "llama3-70b-8192")
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.pydantic_v1 import BaseModel, Field

tagging_prompt = ChatPromptTemplate.from_template(
"""
Extract the desired information from the following passage.

Only extract the properties mentioned in the 'Classification' function.

Passage:
{input}
"""
)

class Classification(BaseModel):
    sentiment: str = Field(...,enum=["happy","neutral","sad"])  # required field ho bhanna lai ... use garya
    political_tendency: str = Field(..., description="Political tendency of the user",enum=["conservative", "liberal","independent"])
    language: str = Field(..., description="The language the text is written in", enum=["English","Nepali"])
```

```python
tagging_chain = tagging_prompt | llmModel.with_structured_output(schema=Classification)
```

```python
kpoli="Kp oli is a very bad prime minister of Nepal"
```

```python
Balen="Balen is very good mayor of nepal"
```

```python
tagging_chain.invoke({"input":kpoli})
```

```python
tagging_chain.invoke({"input":Balen})
```
